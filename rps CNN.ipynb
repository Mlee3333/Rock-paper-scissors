```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers,models
from tensorflow.keras.utils import to_categorical, image_dataset_from_directory
from sklearn.model_selection import train_test_split
```

```python
batchSize = 32
imSize = (128,128)
```

```python
trainPath = "C:\\Users\\morga\\OneDrive\\Desktop\\Jetlearn\\Deep learning\\rockpaperscissors\\rpstrain\\rps"
testPath = "C:\\Users\\morga\\OneDrive\\Desktop\\Jetlearn\\Deep learning\\rockpaperscissors\\rpstest\\rps"

trainDS = tf.keras.utils.image_dataset_from_directory(
    directory = trainPath,
    labels = "inferred",
    label_mode = "categorical",
    batch_size = batchSize,
    image_size = imSize,
    shuffle = True
)
testDS = tf.keras.utils.image_dataset_from_directory(
    directory = testPath,
    labels = "inferred",
    label_mode = "categorical",
    batch_size = batchSize,
    image_size = imSize,
    shuffle = True
)

print(trainDS.class_names)
```

```python
trainImg = []
trainLbl = []
for batchImages, batchLabels in trainDS:
    for image in batchImages:
        trainImg.append(image)
    for label in batchLabels:
        trainLbl.append(label)
Xtrain = np.array(trainImg)
ytrain = np.array(trainLbl)
print(Xtrain.shape)
testImg = []
testLbl = []
for batchImages, batchLabels in testDS:
    for image in batchImages:
        testImg.append(image)
    for label in batchLabels:
        testLbl.append(label)
Xtest = np.array(testImg)
ytest = np.array(testLbl)
print(Xtest.shape)
Xtrain = Xtrain/255
Xtest = Xtest/255
```

```python
CNNmodel = models.Sequential([
    layers.Conv2D(32,(3,3),activation="relu",input_shape=(128,128,3)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64,(3,3),activation="relu"),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(128,(3,3),activation="relu"),
    layers.Flatten(),
    layers.Dense(128,"relu"),
    layers.Dense(3,activation="softmax")
])
CNNmodel.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
fittedM = CNNmodel.fit(Xtrain,ytrain,epochs=5,batch_size = 32, validation_data=(Xtest,ytest))
loss,accuracy = CNNmodel.evaluate(Xtest,ytest,verbose=2)
print(accuracy)
```

```python
CNNmodel.save("rps_cnn.keras")
```
